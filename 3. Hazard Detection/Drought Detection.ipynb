```python
"""
Drought Detection Core Functions

Detects meteorological drought events using Standardized Precipitation Index (SPI)
with threshold-based classification. Caller handles data loading and spatial processing.
"""

import numpy as np
from typing import Union


def detect_drought_events(
    spi_values: np.ndarray,
    threshold: float = -1.5,
    drought_code: int = 2,
    non_drought_code: int = 0
) -> np.ndarray:
    """
    Detect drought events from SPI timeseries using threshold exceedance.
    
    Classifies days as drought when SPI falls below specified threshold.
    NaN values in input propagate as non-drought (non_drought_code) since
    NaN comparisons yield False in numpy.
    
    Parameters
    ----------
    spi_values : np.ndarray
        1D array of SPI values (any timescale)
    threshold : float
        SPI threshold for drought classification (default: -1.5 for severe drought)
    drought_code : int
        Value to assign for drought days (default: 2)
    non_drought_code : int
        Value to assign for non-drought days (default: 0)
    
    Returns
    -------
    np.ndarray
        Integer array with drought classification:
          - drought_code where SPI < threshold
          - non_drought_code otherwise (including NaN positions)
    """
    # Threshold comparison (NaN values yield False → non_drought_code)
    drought_mask = spi_values < threshold
    return np.where(drought_mask, drought_code, non_drought_code).astype(int)


def process_grid_cell_drought(
    spi_timeseries: np.ndarray,
    threshold: float = -1.5
) -> np.ndarray:
    """
    Detect drought events for a single grid cell's SPI timeseries.
    
    Parameters
    ----------
    spi_timeseries : np.ndarray
        1D array of SPI values for one grid cell (time dimension only)
    threshold : float
        SPI threshold for drought classification (default: -1.5)
    
    Returns
    -------
    np.ndarray
        Binary drought occurrence array (2 = drought, 0 = no drought)
    """
    return detect_drought_events(
        spi_values=spi_timeseries,
        threshold=threshold,
        drought_code=2,
        non_drought_code=0
    )
```
