```python
"""
Compound Hazard Detection Functions

Combines heatwave, drought, and fire hazards into a unified classification map
using binary encoding. Caller handles data loading, temporal alignment, and I/O.
"""

import numpy as np
from typing import Union


def combine_hazard_classes(
    heatwave: np.ndarray,
    drought: np.ndarray,
    fire: np.ndarray
) -> np.ndarray:
    """
    Combine three binary hazard classes into unified compound hazard map.
    
    Uses binary encoding scheme:
      0 = no hazard
      1 = heatwave only
      2 = drought only
      3 = fire only
      4 = heatwave + drought
      5 = heatwave + fire
      6 = drought + fire
      7 = heatwave + drought + fire (full compound)
    
    Parameters
    ----------
    heatwave : np.ndarray
        Binary array (1 = heatwave, 0 = no heatwave)
    drought : np.ndarray
        Binary array (2 = drought, 0 = no drought)
    fire : np.ndarray
        Binary array (3 = fire hazard, 0 = no fire)
    
    Returns
    -------
    np.ndarray
        Integer array with compound hazard classification (0-7)
    """
    # Initialize with base hazards
    combined = np.zeros_like(heatwave, dtype=np.int32)
    combined[heatwave == 1] = 1
    combined[drought == 2] = 2
    combined[fire == 3] = 3
    
    # Two-way combinations (order matters: later assignments override earlier)
    combined[(heatwave == 1) & (drought == 2)] = 4
    combined[(heatwave == 1) & (fire == 3)] = 5
    combined[(drought == 2) & (fire == 3)] = 6
    
    # Three-way compound hazard (highest priority)
    combined[(heatwave == 1) & (drought == 2) & (fire == 3)] = 7
    
    return combined


def process_compound_hazards(
    heatwave: np.ndarray,
    drought: np.ndarray,
    fire: np.ndarray
) -> np.ndarray:
    """
    Generate compound hazard classification from individual hazard arrays.
    
    Parameters
    ----------
    heatwave : np.ndarray
        Heatwave occurrence array (1 = heatwave, 0 = no heatwave)
    drought : np.ndarray
        Drought occurrence array (2 = drought, 0 = no drought)
    fire : np.ndarray
        Fire hazard occurrence array (3 = fire hazard, 0 = no fire)
    
    Returns
    -------
    np.ndarray
        Compound hazard classification array (0-7)
    """
    return combine_hazard_classes(heatwave, drought, fire)
```
